## Task 1: Count hashtags
Count every hashtag occurrence in the original tweets, ignoring case. Do this before removing punctuation.

```python
from pathlib import Path
from collections import Counter
import re
import kagglehub
import pandas as pd

path = Path(kagglehub.dataset_download("seriousran/appletwittersentimenttexts"))
print([p.name for p in path.rglob("*.csv")])
```

```python
df = pd.read_csv(path / "apple-twitter-sentiment-texts.csv")
print(df.columns.tolist())
tweets = df["text"].dropna().astype(str)
hashtags = Counter(tag.lower() for text in tweets for tag in re.findall(r"#\w+", text))
print("Total hashtag occurrences:", sum(hashtags.values()))
print("Unique hashtags:", len(hashtags))
print(pd.DataFrame(hashtags.most_common(10), columns=["hashtag", "count"]))
```

## Task 2: Compile a pattern
`re.compile()` creates a reusable regular expression object. It is useful when the same pattern is used for several searches or substitutions.

```python
text = "This year is 2021"
pattern = re.compile(r"\d+")
print(type(pattern))
print(pattern.sub("2022", text))
```

## Task 3: Split with regex
`re.split()` separates a string wherever the pattern matches. Whitespace around the matches remains in the results.

```python
print(re.split(r"\d+", "a 11 b 2 3 c 4"))
```

## Task 4: spaCy and NLTK tokenization
`spacy.load()` loads a trained language pipeline, including its tokenizer. Both tokenizers split this sentence into `I`, `'m`, `enjoying`, `the`, `NLP`, `course`, and `!`; there is no difference for this example.

```python
import spacy
import nltk

nltk.download("punkt_tab", quiet=True)
nlp = spacy.load("en_core_web_sm")
text = "I'm enjoying the NLP course!"
print("spaCy:")
for token in nlp(text):
    print(token.text)
print("NLTK:")
print(nltk.word_tokenize(text))
```
