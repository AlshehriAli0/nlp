```python
import spacy
from spacy import displacy

nlp = spacy.load("en_core_web_sm")
```

## Task 1: POS tags

```python
sen = nlp("I like watching football matches. I do not know how to play it though")
assert sen[12].text == "play"
print("Coarse tag:", sen[12].pos_)
print("Fine tag:", sen[12].tag_)
print("Meaning:", spacy.explain(sen[12].tag_))

for word in sen:
    print(f"{word.text:12} {word.pos_:10} {word.tag_:8} {spacy.explain(word.tag_)}")
```

## Task 2: Same word, different use

```python
sentences = [
    "Did you search on Google?",
    "Let us google it.",
    "The has placed the food on the table",
    "She plans to table a motion for debate in Parliament",
]

for sentence in sentences:
    print(sentence)
    for word in nlp(sentence):
        if word.text.lower() in ["google", "table"]:
            print(word.text, word.pos_, word.tag_, spacy.explain(word.tag_))
    print()
```

Google is a proper noun when it names the search engine, but `google` is a verb
when it means to search online. A table is a noun when it is furniture.
In the last sentence, `table` is a verb meaning to put forward a motion.
The third sentence has a typo in the question; I kept it as given.
