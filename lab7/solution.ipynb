## Task 2: Sentiment classification

```python
from transformers import pipeline, AutoTokenizer, AutoModelForSeq2SeqLM, set_seed

classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english", device=-1, framework="pt")
sentences = [
    "The movie was excellent.",
    "The lecture was difficult to understand.",
    "I enjoyed learning about transformers.",
]
for sentence, result in zip(sentences, classifier(sentences)):
    print(sentence, result["label"], f'{result["score"]:.4f}')
```

## Task 3: Summarization with FLAN-T5

```python
tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-small")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-small")
paragraph = (
    "Natural language processing helps computers understand human language. "
    "It is used for translation, sentiment analysis, and summarization. "
    "Transformers learn contextual representations and support many of these tasks."
)
inputs = tokenizer("Summarize: " + paragraph, return_tensors="pt")
output = model.generate(**inputs, max_new_tokens=60, do_sample=False)
print(tokenizer.decode(output[0], skip_special_tokens=True))
```

A decoder-only model generates a reply one token at a time using the conversation as context. Its causal attention lets each token depend on earlier tokens. This makes it suitable for producing open-ended responses of different lengths. Training on dialogue and instructions helps a chatbot follow requests and maintain a useful conversation.

## Optional challenge: Sampling
Sampling chooses the next token randomly from a probability distribution. Without resetting the random seed, repeated runs can produce different continuations for the same prompt.

```python
generator = pipeline("text-generation", model="distilgpt2", device=-1, framework="pt")
for _ in range(3):
    output = generator("Natural language processing is", max_new_tokens=30,
                       do_sample=True, top_k=50, top_p=0.95,
                       pad_token_id=generator.tokenizer.eos_token_id)
    print(output[0]["generated_text"])
```
